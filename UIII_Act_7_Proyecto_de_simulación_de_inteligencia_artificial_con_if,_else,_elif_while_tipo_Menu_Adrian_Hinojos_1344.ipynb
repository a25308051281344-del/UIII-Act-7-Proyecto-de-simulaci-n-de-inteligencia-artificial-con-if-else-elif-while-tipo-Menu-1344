```python
print("===================================")
print("   IA DISEÑADOR VIRTUAL 🎨🤖")
print("===================================")

while True:

    print("\nMENÚ PRINCIPAL")
    print("1. Diseñar logo")
    print("2. Diseñar página web")
    print("3. Diseñar habitación")
    print("4. Salir")

    opcion = input("Selecciona una opción: ")

    # OPCIÓN 1
    if opcion == "1":

        print("\n--- DISEÑO DE LOGO ---")

        color = input("¿Color favorito? ")

        if color == "rojo":
            print("🤖 IA: Tu logo será moderno y energético.")

        elif color == "azul":
            print("🤖 IA: Tu logo tendrá estilo profesional.")

        elif color == "negro":
            print("🤖 IA: Tu logo será elegante y minimalista.")

        else:
            print("🤖 IA: Tu logo tendrá un diseño creativo único.")

        estilo = input("¿Quieres estilo gamer o elegante? ")

        if estilo == "gamer":
            print("🎮 IA: Agregando luces y letras futuristas.")

        else:
            print("✨ IA: Creando diseño limpio y sofisticado.")

        print("✅ Diseño de logo terminado.")

    # OPCIÓN 2
    elif opcion == "2":

        print("\n--- DISEÑO DE PÁGINA WEB ---")

        tipo = input("¿Página para tienda o escuela? ")

        if tipo == "tienda":
            print("🛒 IA: Creando botones de compra y catálogo.")

        elif tipo == "escuela":
            print("📚 IA: Agregando horarios y materias.")

        else:
            print("🌐 IA: Creando página personalizada.")

        colorweb = input("¿Tema claro u oscuro? ")

        if colorweb == "oscuro":
            print("🌙 IA: Activando modo oscuro moderno.")

        else:
            print("☀️ IA: Activando modo claro elegante.")

        print("✅ Página web diseñada.")

    # OPCIÓN 3
    elif opcion == "3":

        print("\n--- DISEÑO DE HABITACIÓN ---")

        tamano = input("¿Habitación pequeña o grande? ")

        if tamano == "pequeña":
            print("🛏️ IA: Usando muebles compactos.")

        else:
            print("🏠 IA: Agregando muebles grandes y decoración.")

        estilo = input("¿Estilo moderno o clásico? ")

        if estilo == "moderno":
            print("💡 IA: Colocando luces LED y colores neutros.")

        elif estilo == "clasico":
            print("🪑 IA: Agregando muebles de madera.")

        else:
            print("🎨 IA: Diseñando habitación personalizada.")

        print("✅ Habitación diseñada.")

    # OPCIÓN 4
    elif opcion == "4":

        print("\n👋 Gracias por usar IA Diseñador Virtual.")
        print("Programa finalizado.")
        break

    # OPCIÓN INVÁLIDA
    else:
        print("\n❌ Opción no válida.")
```
